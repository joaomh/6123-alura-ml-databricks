```python
# Verificação de drift na tabela de monitoramento
# Este notebook executa uma query para contar registros com drift detectado
# e disponibiliza o resultado como task value para condicionais no job

from pyspark.sql import SparkSession

# Query para contar registros com drift detectado
query = """
SELECT COUNT(*) as total_drift
FROM workspace.default.log_monitoramento_drift
WHERE drift_detectado = true
"""

# Executar a query
result = spark.sql(query)
total_drift = result.collect()[0]['total_drift']

print(f"Total de registros com drift detectado: {total_drift}")

# Disponibilizar o valor para tasks condicionais
dbutils.jobs.taskValues.set(key="total_drift", value=str(total_drift))

if total_drift > 0:
    print("⚠️ ALERTA: Drift detectado! Re-treino será acionado.")
else:
    print("✓ Nenhum drift detectado. Modelo estável.")
```
